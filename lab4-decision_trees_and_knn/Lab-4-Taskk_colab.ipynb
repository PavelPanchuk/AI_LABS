### Создание набора данных

```python
# Создание датафрейма с dummy variables
def create_df(dic, feature_list):
    out = pd.DataFrame(dic)
    out = pd.concat([out, pd.get_dummies(out[feature_list])], axis = 1)
    out.drop(feature_list, axis = 1, inplace = True)
    return out

# Некоторые значения признаков есть в тесте, но нет в трейне и наоборот
def intersect_features(train, test):
    common_feat = list( set(train.keys()) & set(test.keys()))
    return train[common_feat], test[common_feat]
```

```python
features = ['Внешность', 'Алкоголь_в_напитке',
            'Уровень_красноречия', 'Потраченные_деньги']
```

Постройте от руки (или в графическом редакторе) дерево решений для этого набора данных. Дополнительно (для желающих) – можете сделать отрисовку дерева и написать код для построения всего дерева с помощью **GraphViz**.

**Вопрос 1.** Какова энтропия начальной системы ($S_0$)? Под состояниями системы понимаем значения признака "Поедет" – 0 или 1 (то есть всего 2 состояния).

```python
import math

def entropy2(p1: float, p2: float) -> float:
    return -p1*math.log2(p1)-p2*math.log2(p2)

p1 = 3/7
p2 = 4/7
S0 = entropy2(p1, p2)
print("Entropy: ", S0)
```

**Вопрос 2.** Рассмотрим разбиение обучающей выборки по признаку "Внешность\_приятная". Какова энтропия  $S_1$ левой группы, тех, у кого внешность приятная, и правой группы – $S_2$? Каков прирост информации  при данном разбиении (IG)?

```python
S1_p1 = 3 / 4
S1_p2 = 1 / 4
S2_p1 = 1 / 3
S2_p2 = 2 / 3

S1 = entropy2(S1_p1, S1_p2)
S2 = entropy2(S2_p1, S2_p2)
print(f'Beautiful = {S1}\nAwful = {S2}')

IG = S0 - p2 * S1 - p1 * S2
print(f'IG = {IG}')
```

## Часть 2. Функции для расчета энтропии и прироста информации

Пример для проверки: 9 синих шариков и 11 желтых. Пусть шарик находится в состоянии "1", если он синий и "0" – если он желтый.

```python
balls = [1 for i in range(9)] + [0 for i in range(11)]
```

Дальше пусть шарики разбиваются на 2 группы
![png](https://github.com/PavelPanchuk/AI_LABS/blob/main/lab4-decision_trees_and_knn/img/decision_tree4.png?raw=1)

```python
# две группы
balls_left  = [1 for i in range(8)] + [0 for i in range(5)] # 8 синих и 5 желтых
balls_right = [1 for i in range(1)] + [0 for i in range(6)] # 1 синий и 6 желтых
```

Реализуйте функцию для расчета энтропии Шеннона.

```python
def entropy(a_list):

    lst_counts = collections.Counter(a_list)
    entropy_value = 0
    for count in lst_counts.values():
        p = count / len(a_list)
        entropy_value += -p * math.log2(p)
    return entropy_value

# Проверка:
print(entropy(balls)) # 9 синих и 11 желтых
print(entropy(balls_left)) # 8 синих и 5 желтых
print(entropy(balls_right)) # 1 синий и 6 желтых
print(entropy([1,2,3,4,5,6])) # энтропия игральной кости с несмещенным центром тяжести
```

**Вопрос 3.** Чему равна энтропия состояния, заданного списком `balls_left`?

```python
balls_left  = [1 for i in range(8)] + [0 for i in range(5)] # 8 синих и 5 желтых

# функция расчета энтропии Шеннона
def entropy(a_list):
    lst_counts = collections.Counter(a_list)
    entropy_value = 0
    for count in lst_counts.values():
        p = count / len(a_list)
        entropy_value += -p * math.log2(p)
    return entropy_value

print(entropy(balls_left))
```

**Вопрос 4.** Чему равна энтропия "честной" монетки?

```python
# расчет прироста информации

def information_gain(root, left, right):
    ''' root - изначальный набор данных, left и right два разбиения изначального набора'''

    # Ваш код здесь
    return entropy(root) - (len(left)/len(root))*entropy(left) - (len(right)/len(root))*entropy(right)
```

**Вопрос 5.** Каков прирост информации при разделении выборки на `balls_left` и `balls_right`?

```python
print(information_gain(balls, balls_left, balls_right))
```

```python
def best_feature_to_split(X, y):

    best_ig = -1
    best_feature = None
    initial_entropy = entropy(list(y))

    for feature in X.columns:
        ig_feature = 0
        for value in X[feature].unique():
            left_indices = X[feature] == value
            right_indices = X[feature] != value

            y_left = y[left_indices]
            y_right = y[right_indices]

            if len(y_left) > 0 and len(y_right) > 0: # avoid division by zero if one branch is empty
                ig_feature +=  (len(y_left)/len(y)) * entropy(list(y_left)) + (len(y_right)/len(y)) * entropy(list(y_right))

        current_ig = initial_entropy - ig_feature

        if current_ig > best_ig:
            best_ig = current_ig
            best_feature = feature

    return best_feature
best_feature = best_feature_to_split(df_train, y)
print(f"Наилучший признак для разбиения: {best_feature}")
```
